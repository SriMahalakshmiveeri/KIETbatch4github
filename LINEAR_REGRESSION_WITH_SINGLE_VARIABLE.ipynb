```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn import linear_model
```

```python
df=pd.read_csv("/content/canada_per_capita_income.csv")
df
```

```python
%matplotlib inline
plt.xlabel("year")
plt.ylabel("income_(us$)")
plt.scatter(df.year,df['income_(us$)'])
```

```python
reg=linear_model.LinearRegression()
reg.fit(df[['year']],df['income_(us$)'])
```

```python
reg.predict(np.array([[2020]]))
```

```python
reg.coef_
```

```python
reg.intercept_
```

```python
828.46507522*2020+(-1632210.7578554575)
```

```python
%matplotlib inline
plt.xlabel("year")
plt.ylabel("income_(us$)")
plt.scatter(df.year,df['income_(us$)'])
plt.plot(df.year,reg.predict(df[['year']]),color='red')
```
